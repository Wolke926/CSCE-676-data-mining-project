```python
import math
import pandas as pd
import numpy as np
from collections import defaultdict
from tqdm import tqdm


def time_decay(delta_days, half_life=7):
    return 0.5 ** (delta_days / half_life)

def itemcf(
    history_df,
    topk=50,max_user_items=20,item_sim_topk=100,
    start_date=None,user_col="user_id",item_col="item_id",date_col="t_dat",):
    hist = history_df.copy()
    hist[date_col] = pd.to_datetime(hist[date_col])

    # 可选：只用最近一段历史，减少计算量
    if start_date is not None:
        hist = hist[hist[date_col] > pd.to_datetime(start_date)]

    hist = hist.sort_values([user_col, date_col, item_col])

    user_item_time = (
        hist.groupby(user_col)[[item_col, date_col]]
        .apply(lambda x: list(zip(x[item_col], x[date_col]))[-max_user_items:])
    )

    item_sim = defaultdict(lambda: defaultdict(float))
    item_cnt = defaultdict(int)

    # 1. build item-item similarity
    for u, item_time_list in tqdm(user_item_time.items(), desc="Build ItemCF"):
        items = [i for i, _ in item_time_list]
        times = [t for _, t in item_time_list]
        L = len(items)

        for i in range(L):
            item_i = items[i]
            time_i = times[i]
            item_cnt[item_i] += 1

            for j in range(L):
                if i == j:
                    continue

                item_j = items[j]
                time_j = times[j]

                if item_i == item_j:
                    continue

                delta = abs((time_i - time_j).days)

                # 位置越近，权重越大
                loc_weight = 1.0 / (1 + abs(i - j))

                # 时间越近，权重越大
                decay_weight = time_decay(delta)

                item_sim[item_i][item_j] += loc_weight * decay_weight

    # 2. normalize
    for i, related_items in item_sim.items():
        for j in related_items:
            item_sim[i][j] /= math.sqrt(item_cnt[i] * item_cnt[j])

    # 3. keep topk similar items for each item
    item_sim = {
        i: dict(sorted(related.items(), key=lambda x: x[1], reverse=True)[:item_sim_topk])
        for i, related in item_sim.items()
    }

    # 4. recall candidates for each user
    rows = []

    for u, item_time_list in tqdm(user_item_time.items(), desc="Recall ItemCF"):
        user_items = [i for i, _ in item_time_list]
        scores = defaultdict(float)

        for item in user_items:
            if item not in item_sim:
                continue

            for j, sim in item_sim[item].items():
                scores[j] += sim

        rec_items = sorted(scores.items(), key=lambda x: x[1], reverse=True)[:topk]

        for item_id, score in rec_items:
            rows.append([u, item_id, score])

    cand = pd.DataFrame(rows, columns=[user_col, item_col, "itemcf_score"])

    return cand
```

```python
def gen_train_valid_ver2(df, n_windows, n_neg, itemcf, valid_start_date):
    df = df.copy()
    df["t_dat"] = pd.to_datetime(df["t_dat"])
    df = df.sort_values(["t_dat", "user_id", "item_id"])

    valid_start_date = pd.to_datetime(valid_start_date)
    all_items = df["item_id"].unique()
    rng = np.random.default_rng(42)

    train_list = []
    valid_list = []

    for i in range(n_windows + 1):
        # i = 0 valid, i > 0 train
        label_start = valid_start_date - pd.Timedelta(days=7 * i)
        label_end = label_start + pd.Timedelta(days=7)

        label_df = df[
            (df["t_dat"] >= label_start) &
            (df["t_dat"] < label_end)
        ]

        history_df = df[df["t_dat"] < label_start]

        if len(label_df) == 0 or len(history_df) == 0:
            continue

        # 正样本：当前 label week 真实购买
        pos = label_df[["user_id", "item_id"]].drop_duplicates()
        pos["label"] = 1
        pos["window"] = i

        # ItemCF 只用来召回候选，不用 score
        cand = itemcf(history_df)[["user_id", "item_id"]].drop_duplicates()

        # 负样本池：召回了，但当前 label week 没买
        pos_pairs = set(zip(pos["user_id"], pos["item_id"]))
        cand["_pair"] = list(zip(cand["user_id"], cand["item_id"]))
        neg_pool = cand[~cand["_pair"].isin(pos_pairs)].drop(columns="_pair")

        neg_rows = []

        for u, g in pos.groupby("user_id"):
            k = len(g) * n_neg

            user_neg_items = (
                neg_pool[neg_pool["user_id"] == u]["item_id"]
                .drop_duplicates()
                .tolist()
            )

            if len(user_neg_items) >= k:
                neg_items = rng.choice(user_neg_items, size=k, replace=False).tolist()
            else:
                neg_items = user_neg_items.copy()

                bought = set(g["item_id"])
                used = set(neg_items)

                while len(neg_items) < k:
                    item = rng.choice(all_items)
                    if item not in bought and item not in used:
                        neg_items.append(item)
                        used.add(item)

            for item in neg_items:
                neg_rows.append([u, item, 0, i])

        neg = pd.DataFrame(
            neg_rows,
            columns=["user_id", "item_id", "label", "window"]
        )

        samples = pd.concat([pos, neg], ignore_index=True)

        if i == 0:
            valid_list.append(samples)
        else:
            train_list.append(samples)

    train_samples = pd.concat(train_list, ignore_index=True)
    valid_samples = pd.concat(valid_list, ignore_index=True)

    return train_samples, valid_samples
```

```python
import pandas as pd
import numpy as np

def gen_train_valid(df, n_windows, n_neg, itemcf, valid_start_date):
    df = df.copy()
    df["t_dat"] = pd.to_datetime(df["t_dat"])
    df = df.sort_values(["t_dat", "user_id", "item_id"])

    valid_start_date = pd.to_datetime(valid_start_date)
    all_items = df["item_id"].unique()
    rng = np.random.default_rng(42)

    train_list = []
    valid_list = []

    for i in range(n_windows + 1):
        # i = 0 valid, i > 0 train
        label_start = valid_start_date - pd.Timedelta(days=7 * i)
        label_end = label_start + pd.Timedelta(days=7)

        label_df = df[
            (df["t_dat"] >= label_start) &
            (df["t_dat"] < label_end)
        ]

        history_df = df[df["t_dat"] < label_start]

        if len(label_df) == 0 or len(history_df) == 0:
            continue

        # 正样本：这一周真实购买
        pos = label_df[["user_id", "item_id"]].drop_duplicates()
        pos["label"] = 1
        pos["window"] = i

        # 候选：用历史做 itemcf 召回
        cand = itemcf(history_df)[["user_id", "item_id"]].drop_duplicates()

        # 负样本池：召回了，但这一周没买
        pos_pairs = set(zip(pos["user_id"], pos["item_id"]))
        cand["_pair"] = list(zip(cand["user_id"], cand["item_id"]))
        neg_pool = cand[~cand["_pair"].isin(pos_pairs)].drop(columns="_pair")

        # 每个用户采 n_neg * 正样本数 个负样本
        neg_list = []

        for u, g in pos.groupby("user_id"):
            k = len(g) * n_neg

            user_neg = neg_pool[neg_pool["user_id"] == u]["item_id"].unique()

            if len(user_neg) >= k:
                neg_items = rng.choice(user_neg, size=k, replace=False)
            else:
                neg_items = list(user_neg)

                bought = set(g["item_id"])
                while len(neg_items) < k:
                    item = rng.choice(all_items)
                    if item not in bought and item not in neg_items:
                        neg_items.append(item)

            for item in neg_items:
                neg_list.append([u, item, 0, i])

        neg = pd.DataFrame(
            neg_list,
            columns=["user_id", "item_id", "label", "window"])

        samples = pd.concat([pos, neg], ignore_index=True)

        if i == 0:
            valid_list.append(samples)
        else:
            train_list.append(samples)

    train_df = pd.concat(train_list, ignore_index=True)
    valid_df = pd.concat(valid_list, ignore_index=True)

    X_train = train_df[["user_id", "item_id", "window"]]
    y_train = train_df["label"]

    X_valid = valid_df[["user_id", "item_id", "window"]]
    y_valid = valid_df["label"]

    return X_train, y_train, X_valid, y_valid
```

```python
import logging
from gensim.models import Word2Vec
import pickle
from tqdm import tqdm
import sys
from pathlib import Path
from gensim.models import Word2Vec
import pandas as pd
import numpy as np

model_dir = Path("models")
data_dir = Path('data')
#should sort first, sequence is important
inter = embd_data.sort_values(['user_id','t_dat','item_id']).reset_index(drop = True)

#every sentence is a list of articles for every customer,use skip-gram to calculate embd for every article
seq_list = inter.groupby(['user_id'])['item_id'].apply(lambda x: [str(id) for id in x]).to_list()
#sg=1 skip-gram, sg = 0, cbow
logging.basicConfig(
    format='%(asctime)s : %(levelname)s : %(message)s',
    level=logging.INFO)
model_sg = Word2Vec(seq_list, vector_size = 64, window = 10, min_count = 2, sg = 1, sample = 1e-3, negative = 15,workers=8,seed = 1,epochs = 5)
model_sg.save(str(model_dir/'item_id_sg_model.model'))
```

```python
model_sg = pickle.load(open(model_dir/'item_id_sg_model.model','rb'))
model_sg.wv.save(str(model_dir / 'item_id_sg.wordvectors'))
```

```python
# use the model above, we can use the average of embd of bought articles for customer
num_users = len(ALL_CUSTOMERS)
num_items = len(ALL_ARTICLES)

emb_dim = model_sg.wv.vector_size

customer_embd = np.zeros((num_users, emb_dim), dtype=np.float32)

feed_id_seq = (
    inter.groupby('user_id')['item_id']
    .apply(list)
    .reset_index()
)

for uid, items in tqdm(feed_id_seq.values):

    valid_items = [str(i) for i in items if str(i) in model_sg.wv]
    if not valid_items:
        continue
# the user embd is the average of embds of items they bought
    vec = model_sg.wv[valid_items].mean(axis=0)

    norm = np.linalg.norm(vec)
    if norm > 0:
        vec = vec / norm

    customer_embd[int(uid)] = vec.astype(np.float32)

np.save('w2v_sg_user_embd.npy', customer_embd)
```

```python
emb_dim = model_sg.wv.vector_size

# row_number = item_id
item_embd = np.zeros((num_items, emb_dim), dtype=np.float32)

for item_id in tqdm(range(num_items)):
    key = str(item_id)

    if key in model_sg.wv:
        vec = model_sg.wv[key]
        norm = np.linalg.norm(vec)
        if norm > 0:
            vec = vec / norm
        item_embd[item_id] = vec.astype(np.float32)

np.save('w2v_sg_item_embd.npy', item_embd)
```

```python
def build_dnn_model(customer_embd, item_embd, trainable_emb=True):
    num_users, emb_dim = customer_embd.shape
    num_items, item_emb_dim = item_embd.shape

    assert emb_dim == item_emb_dim, "user embedding dim and item embedding dim must match"

    user_input = layers.Input(shape=(), name="user_id", dtype=tf.int32)
    item_input = layers.Input(shape=(), name="item_id", dtype=tf.int32)

    user_embedding_layer = layers.Embedding(
        input_dim=num_users,
        output_dim=emb_dim,
        weights=[customer_embd],
        trainable=trainable_emb,
        name="user_embedding"
    )

    item_embedding_layer = layers.Embedding(
        input_dim=num_items,
        output_dim=emb_dim,
        weights=[item_embd],
        trainable=trainable_emb,
        name="item_embedding"
    )

    user_vec = user_embedding_layer(user_input)
    item_vec = item_embedding_layer(item_input)

    # shape: [batch, emb_dim]
    user_vec = layers.Flatten()(user_vec)
    item_vec = layers.Flatten()(item_vec)

    interaction = layers.Multiply()([user_vec, item_vec])

    x = layers.Concatenate()([
        user_vec,
        item_vec,
        interaction
    ])

    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.2)(x)

    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.2)(x)

    x = layers.Dense(64, activation="relu")(x)

    output = layers.Dense(1, activation="sigmoid", name="pred")(x)

    model = Model(
        inputs={
            "user_id": user_input,
            "item_id": item_input,
        },
        outputs=output
    )

    return model
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
```

```python
model = build_dnn_model(
    customer_embd=customer_embd,
    item_embd=item_embd,
    trainable_emb=True
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.AUC(name="auc"),
        tf.keras.metrics.BinaryAccuracy(name="acc")
    ]
)

model.summary()
```

```python
callbacks = [
    EarlyStopping(
        monitor="val_auc",
        mode="max",
        patience=2,
        restore_best_weights=True
    ),
    ModelCheckpoint(
        "dnn_recall_rerank.keras",
        monitor="val_auc",
        mode="max",
        save_best_only=True
    )
]
```

```python
history = model.fit(
    train_ds,
    validation_data=valid_ds,
    epochs=10,
    callbacks=callbacks
)
```
